```python
# 1. Install dependencies
!pip install ultralytics

import cv2
from ultralytics import YOLO
from google.colab.patches import cv2_imshow
from IPython.display import clear_output

# 2. Load the YOLOv8 model
model = YOLO('yolov8n.pt')

# 3. Path to your video (Ensure you upload the file to Colab first)
video_path = "/026c7465-309f6d33.mp4"
cap = cv2.VideoCapture(video_path)

# Loop through the video frames
while cap.isOpened():
    success, frame = cap.read()

    if success:
        # 4. Run YOLOv8 tracking (Same logic as your code)
        # results is a list, so we take results[0]
        results = model.track(frame, persist=True, tracker="botsort.yaml")

        # 5. Visualize the results
        annotated_frame = results[0].plot()

        # 6. Colab-specific display logic
        clear_output(wait=True) # Clears the previous frame for a smooth video look
        cv2_imshow(annotated_frame) # Colab's replacement for cv2.imshow

        # Optional: Break if you want to stop early (manual stop in Colab)
    else:
        break

cap.release()
cv2.destroyAllWindows()
```
